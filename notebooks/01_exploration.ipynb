# Exploration de MovieLens 1M
Ce notebook explore les distributions des ratings, utilisateurs et films.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

users_cols = ['user_id', 'gender', 'age', 'occupation', 'zip']
users = pd.read_csv('../data/raw/ml-1m/users.dat', sep='::', engine='python', names=users_cols, encoding='latin-1')

movies_cols = ['movie_id', 'title', 'genres']
movies = pd.read_csv('../data/raw/ml-1m/movies.dat', sep='::', engine='python', names=movies_cols, encoding='latin-1')

ratings_cols = ['user_id', 'movie_id', 'rating', 'timestamp']
ratings = pd.read_csv('../data/raw/ml-1m/ratings.dat', sep='::', engine='python', names=ratings_cols, encoding='latin-1')

print(f"Users: {len(users)}, Movies: {len(movies)}, Ratings: {len(ratings)}")
```

```python
# Distribution des ratings
sns.countplot(x='rating', data=ratings)
plt.title('Distribution des Ratings')
plt.show()
```

```python
# Nombre de notes par film
ratings_per_movie = ratings.groupby('movie_id').size()
sns.histplot(ratings_per_movie, bins=50)
plt.title('Distribution du nombre de notes par film')
plt.yscale('log')
plt.show()
```
